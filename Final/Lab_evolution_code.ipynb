```python
import numpy as np

# Taking user input
x = np.array(list(map(float, input("Enter values of x (space-separated): ").split())))
y = np.array(list(map(float, input("Enter values of f(x) (space-separated): ").split())))
xp = float(input("Enter the value of x where derivative is needed: "))

n = len(x)
h = x[1] - x[0]

# Construct backward difference table
diff_table = np.zeros((n, n))
diff_table[:, 0] = y

for j in range(1, n):
    for i in range(j, n):
        diff_table[i][j] = diff_table[i][j - 1] - diff_table[i - 1][j - 1]

# Print Backward Difference Table
print("\nBackward Difference Table:")
header = f"{'x':<8}{'f(x)':<10}" + "".join([f"∇^{j}y".ljust(10) for j in range(1, n)])
print(header)
print("-" * len(header))

for i in range(n):
    row = f"{x[i]:<8.2f}"
    for j in range(i + 1):
        row += f"{diff_table[i][j]:<10.2f}"
    print(row)

# Calculate u and first derivative dynamically for any number of points
u = (xp - x[-1]) / h

# General coefficients for Newton's Backward First Derivative up to 4th order
coeffs = [
    1,
    (2 * u + 1) / 2,
    (3 * u**2 + 6 * u + 2) / 6,
    (4 * u**3 + 18 * u**2 + 22 * u + 6) / 24,
]

derivative = 0
for j in range(1, min(n, 5)):
    derivative += coeffs[j - 1] * diff_table[n - 1][j]

derivative /= h

print(f"\nValue of u at x = {xp}: {u:.4f}")
print(f"First derivative f'({xp}) = {derivative:.4f}")
```

```python
import numpy as np

# Taking user input
x = np.array(list(map(float, input("Enter values of x (space-separated): ").split())))
y = np.array(list(map(float, input("Enter values of y (space-separated): ").split())))
xp_input = input("Enter a value of x to estimate y (or press Enter to skip): ").strip()

n = len(x)

# Calculate required columns
x2 = x**2
x3 = x**3
x4 = x**4
xy = x * y
x2y = (x**2) * y

# Print Calculation Table
print("\nCalculation Table:")
print(f"{'x':<8}{'y':<8}{'x^2':<10}{'x^3':<10}{'x^4':<10}{'xy':<10}{'x^2y':<10}")
print("-" * 66)
for i in range(n):
    print(f"{x[i]:<8.2f}{y[i]:<8.2f}{x2[i]:<10.2f}{x3[i]:<10.2f}{x4[i]:<10.2f}{xy[i]:<10.2f}{x2y[i]:<10.2f}")
print("-" * 66)
print(f"{'Sum:':<8}{np.sum(y):<8.2f}{np.sum(x2):<10.2f}{np.sum(x3):<10.2f}{np.sum(x4):<10.2f}{np.sum(xy):<10.2f}{np.sum(x2y):<10.2f}")
print(f"(Sum of x = {np.sum(x):.2f}, n = {n})\n")

# Normal Equations Matrix: A * [a, b, c]^T = B
A = np.array([
    [n,          np.sum(x),  np.sum(x2)],
    [np.sum(x),  np.sum(x2), np.sum(x3)],
    [np.sum(x2), np.sum(x3), np.sum(x4)]
])

B = np.array([
    np.sum(y),
    np.sum(xy),
    np.sum(x2y)
])

# Solve for a, b, c
a, b, c = np.linalg.solve(A, B)

print("Coefficients:")
print(f"a = {a:.4f}")
print(f"b = {b:.4f}")
print(f"c = {c:.4f}")
print(f"\nFitted Equation: y = {a:.4f} + {b:.4f}x + {c:.4f}x^2")

# Evaluate at the extra x value if provided
if xp_input:
    xp = float(xp_input)
    yp = a + b * xp + c * (xp**2)
    print(f"Estimated value at x = {xp}: y = {yp:.4f}")
```
